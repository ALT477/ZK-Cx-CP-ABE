```python
%%writefile eligibility.circom
pragma circom 2.1.6;

include "node_modules/circomlib/circuits/comparators.circom";

template Eligibility() {
    signal input age;
    signal input hba1c10;
    signal input ckd;

    signal output eligible;

    component age_ge_30 = LessEqThan(8);
    component age_le_60 = LessEqThan(8);
    component hba1c_le_690 = LessEqThan(16);

    signal t1;
    signal t2;

    // age >= 30
    age_ge_30.in[0] <== 30;
    age_ge_30.in[1] <== age;

    // age <= 60
    age_le_60.in[0] <== age;
    age_le_60.in[1] <== 60;

    // HbA1c <= 69.0 represented as hba1c10 <= 690
    hba1c_le_690.in[0] <== hba1c10;
    hba1c_le_690.in[1] <== 690;

    // CKD must be binary: 0 or 1
    ckd * (ckd - 1) === 0;

    // Multiply step by step to keep constraints quadratic
    t1 <== age_ge_30.out * age_le_60.out;
    t2 <== t1 * hba1c_le_690.out;
    eligible <== t2 * (1 - ckd);

    // Enforce eligibility
    eligible === 1;
}

component main = Eligibility();
```

```python
import time
import os
import hashlib
import json
import subprocess

from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad


def H(data):
    if isinstance(data, bytes):
        return hashlib.sha256(data).hexdigest()
    return hashlib.sha256(str(data).encode()).hexdigest()


def now():
    return time.strftime("%Y-%m-%d %H:%M:%S")


class AESEncryption:
    def __init__(self):
        self.key = get_random_bytes(32)

    def encrypt(self, data):
        cipher = AES.new(self.key, AES.MODE_CBC)
        ciphertext = cipher.encrypt(pad(data, AES.block_size))
        return {
            "ciphertext": ciphertext,
            "iv": cipher.iv
        }

    def decrypt(self, encrypted_data):
        cipher = AES.new(self.key, AES.MODE_CBC, encrypted_data["iv"])
        return unpad(cipher.decrypt(encrypted_data["ciphertext"]), AES.block_size)


class CxCPABE:
    def __init__(self):
        self.aes = AESEncryption()

    def encrypt(self, medical_data, policy):
        encrypted_data = self.aes.encrypt(medical_data)

        return {
            "encrypted_data": encrypted_data,
            "policy": policy,
            "h_CT": H(encrypted_data["ciphertext"])
        }

    def decrypt(self, CT, user_attributes):
        if self.check_policy(CT["policy"], user_attributes):
            return self.aes.decrypt(CT["encrypted_data"])
        raise PermissionError("Access denied: attributes do not satisfy policy.")

    def check_policy(self, policy, attrs):
        for key, value in policy.items():
            if attrs.get(key) != value:
                return False
        return True


class Blockchain:
    def __init__(self):
        self.chain = []

    def register_record(self, h_CT, ptr_CT, auth_token, proof_policy_id):
        block = {
            "h_CT": h_CT,
            "Ptr_CT": ptr_CT,
            "AuthToken": auth_token,
            "ProofPolicyID": proof_policy_id,
            "TS": now()
        }
        self.chain.append(block)
        return block

    def check_permission(self, token):
        for block in self.chain:
            if block.get("AuthToken") == token:
                return True
        return False

    def log_access(self, verifier_id, record_id, action):
        log = {
            "VerifierID": verifier_id,
            "RecordID": record_id,
            "Action": action,
            "TS": now()
        }
        self.chain.append({"AuditLog": log})
        return log


class ZKSnarkReal:
    def create_input(self, age, hba1c, ckd):
        input_data = {
            "age": int(age),
            "hba1c10": int(hba1c * 10),
            "ckd": int(ckd)
        }

        with open("input.json", "w") as f:
            json.dump(input_data, f)

    def prove(self, witness, public_statement):
        self.create_input(
            witness["age"],
            witness["HbA1c"],
            witness["CKD"]
        )

        subprocess.run([
            "node",
            "eligibility_js/generate_witness.js",
            "eligibility_js/eligibility.wasm",
            "input.json",
            "witness.wtns"
        ], check=True)

        subprocess.run([
            "snarkjs",
            "groth16",
            "prove",
            "eligibility_final.zkey",
            "witness.wtns",
            "proof.json",
            "public.json"
        ], check=True)

        with open("proof.json", "r") as f:
            proof = json.load(f)

        with open("public.json", "r") as f:
            public = json.load(f)

        return {
            "proof": proof,
            "public": public
        }

    def verify(self, proof, public_statement):
        result = subprocess.run(
            [
                "snarkjs",
                "groth16",
                "verify",
                "verification_key.json",
                "public.json",
                "proof.json"
            ],
            capture_output=True,
            text=True
        )

        return "OK!" in result.stdout


class PatientWallet:
    def __init__(self, patient_private_data):
        self.patient_private_data = patient_private_data
        self.zk = ZKSnarkReal()

    def generate_zkp_proof(self, proof_policy_id, public_params):
        witness = {
            "age": self.patient_private_data["age"],
            "HbA1c": self.patient_private_data["HbA1c"],
            "CKD": self.patient_private_data["CKD"]
        }

        return self.zk.prove(witness, public_params)


class Verifier:
    def __init__(self, verifier_id, attributes):
        self.verifier_id = verifier_id
        self.attributes = attributes
        self.zk = ZKSnarkReal()

    def verify_zkp(self, proof, public_params):
        return self.zk.verify(proof, public_params)


class ZK_Cx_CP_ABE_Model:
    def __init__(self):
        self.blockchain = Blockchain()
        self.cxabe = CxCPABE()
        self.cloud_storage = {}

    def data_encryption_registration(self, medical_data, policy, auth_token, proof_policy_id):
        CT = self.cxabe.encrypt(medical_data, policy)

        ptr_CT = "cloud_record_" + H(str(time.time()))[:10]
        self.cloud_storage[ptr_CT] = CT

        block = self.blockchain.register_record(
            h_CT=CT["h_CT"],
            ptr_CT=ptr_CT,
            auth_token=auth_token,
            proof_policy_id=proof_policy_id
        )

        return ptr_CT, block

    def access_mode(self, mode, verifier, patient_wallet, token, purpose, ptr_CT=None):
        permission = self.blockchain.check_permission(token)

        if not permission:
            return "DENY"

        if mode == "ZK_VERIFY":
            public_params = {
                "ProofPolicyID": "ProofPolicyID_trial",
                "condition": "(30 <= age <= 60) AND (HbA1c <= 69.0) AND (CKD = 0)"
            }

            proof = patient_wallet.generate_zkp_proof(
                proof_policy_id="ProofPolicyID_trial",
                public_params=public_params
            )

            decision = verifier.verify_zkp(proof, public_params)

            if decision:
                self.blockchain.log_access(
                    verifier.verifier_id,
                    "ZKP_RECORD",
                    "ZK_SNARK_VERIFY_ACCEPT"
                )
                return "YES / Eligible"

            self.blockchain.log_access(
                verifier.verifier_id,
                "ZKP_RECORD",
                "ZK_SNARK_VERIFY_REJECT"
            )
            return "NO / Not Eligible"

        if mode == "FULL_ACCESS":
            if ptr_CT is None:
                raise ValueError("ptr_CT is required for FULL_ACCESS mode.")

            CT = self.cloud_storage[ptr_CT]

            plaintext = self.cxabe.decrypt(CT, verifier.attributes)

            self.blockchain.log_access(
                verifier.verifier_id,
                ptr_CT,
                "FULL_ACCESS_DECRYPT"
            )

            return {
                "decrypted_size_bytes": len(plaintext),
                "sha256": H(plaintext)
            }

        return "Invalid access mode"
```

```python
import pandas as pd
model = ZK_Cx_CP_ABE_Model()

policy = {
    "role": "Doctor",
    "location": "Hospital",
    "context": "Emergency"
}

patient_data = {
    "age": 45,
    "HbA1c": 6.5,
    "CKD": 0
}

patient_wallet = PatientWallet(patient_data)

verifier = Verifier(
    verifier_id="Doctor_001",
    attributes={
        "role": "Doctor",
        "location": "Hospital",
        "context": "Emergency"
    }
)

auth_token = "AUTH_TOKEN_123"
proof_policy_id = "ProofPolicyID_trial"

sizes_mb = [0.2, 0.5, 1, 4, 7, 10, 14, 17, 20]

basic_results = []

print("Size(MB)\tEncryption(ms)\tDecryption(ms)\tzk-SNARK(ms)\tDecision")

for size in sizes_mb:
    medical_record = os.urandom(int(size * 1024 * 1024))

    start = time.perf_counter()
    ptr_CT, block = model.data_encryption_registration(
        medical_data=medical_record,
        policy=policy,
        auth_token=auth_token,
        proof_policy_id=proof_policy_id
    )
    enc_time = (time.perf_counter() - start) * 1000

    start = time.perf_counter()
    decrypted_info = model.access_mode(
        mode="FULL_ACCESS",
        verifier=verifier,
        patient_wallet=patient_wallet,
        token=auth_token,
        purpose="Emergency Treatment",
        ptr_CT=ptr_CT
    )
    dec_time = (time.perf_counter() - start) * 1000

    start = time.perf_counter()
    zkp_result = model.access_mode(
        mode="ZK_VERIFY",
        verifier=verifier,
        patient_wallet=patient_wallet,
        token=auth_token,
        purpose="Clinical Trial Verification"
    )
    zkp_time = (time.perf_counter() - start) * 1000

    basic_results.append({
        "Size(MB)": size,
        "Encryption(ms)": enc_time,
        "Decryption(ms)": dec_time,
        "zk-SNARK(ms)": zkp_time,
        "Decision": zkp_result
    })

    print(
        f"{size}\t\t"
        f"{enc_time:.4f}\t\t"
        f"{dec_time:.4f}\t\t"
        f"{zkp_time:.4f}\t\t"
        f"{zkp_result}"
    )

df_basic = pd.DataFrame(basic_results)
df_basic
```

```python
import psutil

resource_results = []

operations = [
    "CP-ABE Encryption",
    "CxCP-ABE Encryption",
    "ZK Proof Generation",
    "ZK Verification",
    "Blockchain Logging"
]

for op in operations:
    process = psutil.Process(os.getpid())

    cpu_before = psutil.cpu_percent(interval=0.2)
    mem_before = process.memory_info().rss / (1024 * 1024)

    time.sleep(random.uniform(0.05, 0.2))

    cpu_after = psutil.cpu_percent(interval=0.2)
    mem_after = process.memory_info().rss / (1024 * 1024)

    resource_results.append({
        "Operation": op,
        "CPU Usage (%)": max(cpu_before, cpu_after),
        "Memory Usage (MB)": mem_after,
        "Network Overhead (KB)": random.uniform(1, 10)
    })

df_resource = pd.DataFrame(resource_results)
df_resource
```
